Exercise 2: Calculating Concentration Using a Calibration Equation
Problem Statement:

In a colorimetric biosensor experiment, the measured absorbance of a glucose sample is 0.45.

The calibration curve is described by:

A = slope × C + intercept

Given:
- Absorbance (A) = 0.45
- Slope = 0.004
- Intercept = 0.05
Write a Python program that calculates the glucose concentration using the calibration equation and displays the result in mg/dL.

```python
absorbance = 0.45
slop = 0.004
intercept = 0.05
concentration = (absorbance - intercept)/slop
print("concentration = ", concentration, "mg/dl")
```

Exercise 5: Calculating Percentage Error in Biosensor Measurements
Problem Statement:

In a glucose biosensor experiment, the actual glucose concentration is 100 mg/dL, while the biosensor measures 106 mg/dL.

Write a Python program that:
1. Stores the actual and measured glucose concentrations in separate variables.
2. Calculates the absolute percentage error using the following formula:
   
   Error (%) = |Actual - Measured| / Actual × 100
3. Displays the percentage error.

```python
Actual = 100
measured = 106
error_percent = abs(Actual-measured)/Actual*100
print("error = ", error_percent, "percent")
```

Exercise 6: Calculating Glucose Concentrations, Average Concentration, and Concentration Range
Problem Statement:

In a colorimetric glucose biosensor experiment, the absorbance of three samples is measured using UV-Vis spectroscopy.

The calibration equation is:

A = slope × C + intercept

Given:
- Slope = 0.005
- Intercept = 0.05
- Sample 1 Absorbance = 0.35
- Sample 2 Absorbance = 0.55
- Sample 3 Absorbance = 0.75
Write a Python program that:
1. Calculates the glucose concentration of each sample in mg/dL.
2. Calculates the average concentration of the three samples.
3. Determines the difference between the highest and lowest concentrations using max() and min().
4. Displays all calculated results.

```python
# calibration value
slope = 0.005
intercept = 0.05

# absorbance value
A1 = 0.35
A2 = 0.55
A3 = 0.75

# concentration value
c1 = (A1-intercept)/slope
c2 = (A2-intercept)/slope
c3 = (A3-intercept)/slope

# calculate average concentration
c_avg = (c1+c2+c3)/3

# calculate concentration range
diffrence = max(c1,c2,c3)-min(c1,c2,c3)

print("c1 = ", c1, "mg/dl")
print("c2 = ", c2, "mg/dl")
print("c3 = ", c3, "mg/dl")
print ("c_avg = ", c_avg, "mg/dl")
print("diffrence = ", diffrence, "mg/dl")
```
